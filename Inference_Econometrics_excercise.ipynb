## Ejercicio 1.-

1.	Considera la siguiente hipótesis nula:

$$H_0:β_2=β_3=...=β_k=0$$

Muestra que, en este caso particular, el estadístico F puede calcularse en términos de la $R^2$ del modelo no restringido:

$$F_{K-1,n-k}=\frac{\frac{R^{2}}{(k-1)}}{\frac{(1-R^{2})}{(n-k)}}$$

El estadístico F con q restricciones lineales se define como:

$$F=\frac{\frac{SSR_r-SSR_{nr}}{q}}{\frac{SSR_{nr}}{(n-k)}}$$

dónde:

*   $SSR_r=$ Sumade Cuadrados de Residuos del **modelo restringido**.
*   $SSR_{nr}=$ Sumade Cuadrados de Residuos del **modelo no restringido**.
*   $q$ = Número de restricciones (grados de libertad del numerador).
*   $n$ = Número de observaciones.
*   $k$ = Número de parámetros en el modelo no restringido (incluyendo el intercepto).
*   $(n - k)$ = Grados de libertad del denominador.

Y sean los modelos:

- **M. No Restringido:**
$$Y=β_1+β_2X_2+...+β_lX_k+u$$

Donde:

*   La suma de cuadrados de sus residuos es $SSR_{nr}$.
*   El $R^2$ de este modelo (que es el $R^2$ en tu fórmula) se define como: $R^2 = 1 - \frac{SSR_{ur}}{SST}$
*   $SST$ (Suma Total de Cuadrados) se define como $\sum(Y_i - \bar{Y})^2$.

y
- **M. Restringido**
dónde:
$$H_0:β_2=...=β_k=0 → Y=β_1+u$$

y.

*   Este es un modelo que solo incluye un intercepto. La estimación de Mínimos Cuadrados Ordinarios (MCO) para $\beta_1$ en este modelo es simplemente la media de Y ($\hat{\beta}_1 = \bar{Y}$).
*   Los residuos de este modelo restringido son $(Y_i - \bar{Y})$.
*   La suma de cuadrados de residuos del modelo restringido ($SSR_r$) es: $SSR_r = \sum(Y_i - \bar{Y})^2$.
*   Por definición, $\sum(Y_i - \bar{Y})^2$ es la Suma Total de Cuadrados ($SST$).
*   Entonces, llegamos a una conclusión clave: $SSR_r = SST$.
*   Ademas que $q=k-1$ por la hipótesis nula.

Sustituyendo en la formula de F se vería.

$$F=\frac{\frac{SSR_r-SSR_{nr}}{k-1}}{\frac{SSR_{nr}}{(n-k)}}$$

Ahora recordemos que $R^2 = \frac{SST - SSR_{ur}}{SST}$ por lo tanto convenientemente $(SST - SSR_{ur}) = R^2 \cdot SST$

Y, sabemos que $1 - R^2 = \frac{SSR_{ur}}{SST}$. Si despejamos, obtenemos:
$SSR_{ur} = (1 - R^2) \cdot SST$

Entonces la formula de F ahora se ve así.

$$F=\frac{\frac{R^{2}SST}{k-1}}{\frac{(1-R^{2})SST}{(n-k)}} → \frac{\frac{R^{2}}{k-1}}{\frac{(1-R^{2})}{(n-k)}} \text{LQQD}$$

## Ejercicio 5.-

Considera el siguiente modelo.

$$Y_t=β_1+β_2X_{2i}+β_3X_{3i}+u_i$$

Y el resultado es:

$$\hat{Y}_t=-0.0636+1.44X_{2i}-0.4898X_{3i}$$

Sabiendo que:

image.png


y que $σ^{2}_e=3.1799$, contrasta la sigueinte hipótesis nula con $α=0.05$:

$$β_1+3β_3=2$$

$$β_3=1$$

Contra la alternativa $H_1:H_0$ es falso

Consideremos la formula general para el estadistico F para restricción lineal.
Que sería:

$$F = \frac{(R\hat{\beta} - r)' [R (X'X)^{-1} R']^{-1} (R\hat{\beta} - r) / q}{\hat{\sigma}^2}$$

-  $\hat{\beta}$ es el coeficiente estimado.

-  $(X'X)^{-1}$ es la matriz de varianza-covarianza.

-  $\hat{\sigma}^2$ es la varianza estimada del error.

-  $R$ es la matriz que define las restricciones.

- $r$ es el vector de valores de las restricciones.

- $q$ es el número de restricciones.

Por el problema sabemos quienes son las $\hat{\beta}$.

$$\hat{\beta} = \begin{pmatrix} \hat{\beta}_1 \\ \hat{\beta}_2 \\ \hat{\beta}_3 \end{pmatrix} = \begin{pmatrix} -0.0636 \\ 1.44 \\ -0.4898 \end{pmatrix}$$

Matriz $R$ y vector $r$: Convertimos la $H_0$ a la forma matricial $R\beta = r$.
$$  R = \begin{pmatrix}
1 & 0 & 3 \\
0 & 0 & 1
\end{pmatrix}
\quad y \quad
r = \begin{pmatrix}
2 \\ 1
\end{pmatrix}
$$

1. Calcular $(R\hat{\beta} - r)$:
$$
R\hat{\beta} = \begin{pmatrix} 1 & 0 & 3 \\ 0 & 0 & 1 \end{pmatrix} \begin{pmatrix} -0.0636 \\ 1.44 \\ -0.4898 \end{pmatrix} = \begin{pmatrix} (1)(-0.0636) + (3)(-0.4898) \\ (1)(-0.4898) \end{pmatrix} = \begin{pmatrix} -1.533 \\ -0.4898 \end{pmatrix}
$$
$$
(R\hat{\beta} - r) = \begin{pmatrix} -1.533 \\ -0.4898 \end{pmatrix} - \begin{pmatrix} 2 \\ 1 \end{pmatrix} = \begin{pmatrix} -3.533 \\ -1.4898 \end{pmatrix}
$$

**2. Calcular la matriz $[R (X'X)^{-1} R']$:**

* Primero, $R(X'X)^{-1}$:
$$
\begin{pmatrix} 1 & 0 & 3 \\ 0 & 0 & 1 \end{pmatrix} \begin{pmatrix} 0.1011 & -0.0007 & -0.0005 \\ -0.0007 & 0.0231 & -0.0162 \\ -0.0005 & -0.0162 & 0.0122 \end{pmatrix}
=
\begin{pmatrix} 0.0996 & -0.0493 & 0.0361 \\ -0.0005 & -0.0162 & 0.0122 \end{pmatrix}
$$

* Ahora, multiplicamos por $R'$
$$
\begin{pmatrix} 0.0996 & -0.0493 & 0.0361 \\ -0.0005 & -0.0162 & 0.0122 \end{pmatrix} \begin{pmatrix} 1 & 0 \\ 0 & 0 \\ 3 & 1 \end{pmatrix} =
\begin{pmatrix} 0.2079 & 0.0361 \\ 0.0361 & 0.0122 \end{pmatrix}
$$

3. Calcular la inversa $[R (X'X)^{-1} R']^{-1}$:Usamos la fórmula de la inversa de una matriz 2x2: $\begin{pmatrix} a & b \\ c & d \end{pmatrix}^{-1} = \frac{1}{ad-bc} \begin{pmatrix} d & -b \\ -c & a \end{pmatrix}$.

Determinante ($ad-bc$): $(0.2079)(0.0122) - (0.0361)^2 = 0.002536 - 0.001303 = 0.001233$

Inversa:
$$
\frac{1}{0.001233} \begin{pmatrix} 0.0122 & -0.0361 \\ -0.0361 & 0.2079 \end{pmatrix}
\approx
\begin{pmatrix} 9.894 & -29.278 \\ -29.278 & 168.61 \end{pmatrix}
$$

4. Calcular el numerador completo: $(R\hat{\beta} - r)' [R(X'X)^{-1}R']^{-1} (R\hat{\beta} - r)$

$$\begin{pmatrix} -3.533 & -1.4898 \end{pmatrix} \begin{pmatrix} 9.894 & -29.278 \\ -29.278 & 168.61 \end{pmatrix} \begin{pmatrix} -3.533 \\ -1.4898 \end{pmatrix} $$

* Multiplicando los dos últimos términos (Matriz Inversa $\times$ vector):
$$
\begin{pmatrix} (9.894)(-3.533) + (-29.278)(-1.4898) \\ (-29.278)(-3.533) + (168.61)(-1.4898) \end{pmatrix}
=
\begin{pmatrix} -34.956 + 43.646 \\ 103.435 - 251.20 \end{pmatrix}
=
\begin{pmatrix} 8.69 \\ -147.765 \end{pmatrix}
$$

* Multiplicando por el vector transpuesto:
$$
\begin{pmatrix} -3.533 & -1.4898 \end{pmatrix}
\begin{pmatrix} 8.69 \\ -147.765 \end{pmatrix}
= (-3.533)(8.69) + (-1.4898)(-147.765) = -30.70 + 220.14 = 189.44
$$

5. Calcular el estadístico F:
$$F = \frac{\text{Numerador} / q}{\hat{\sigma}^2} = \frac{189.51 / 2}{3.1799} = \frac{94.755}{3.1799} \approx \textbf{29.798}$$

```r
F_estadistico <- 29.798
alpha <- 0.05
df1 <- 2
n <- c(30,50,100,200,1000) # El ejercicio no especifica las observaciones, ajustar, asi que hare con varias
df2 <- n - 3

p_value <- pf(F_estadistico, df1, df2, lower.tail = FALSE)

print(paste("F calculado:", F_estadistico))
print(paste("p-value:", p_value))

ifelse(p_value < alpha, "Rechazar H0", "Aceptar H0")
```

Esto significa que no hay evidencia estadística para sostener que ambas restricciones ($\beta_1 + 3\beta_3 = 2$ y $\beta_3 = 1$) sean verdaderas al mismo tiempo.

## Ejercicio 7.-

Demuestra que un estadístico $F_{(1,n-k)}$ es equivalente a un estadpistico $t^{2}_{(n-k)}$

Recordemos que la distribución $t$ (t de Student):
$$t_v = \frac{Z}{\sqrt{\chi^2_v / v}}$$

Donde $v$ son mis grados de libertad.

Ademas que  la distribución $F$ (F de Snedecor):
Una v.a. $F$ con $J$ grados de libertad en el numerador y $v$ grados de libertad en el denominador

$$F_{(J, v)} = \frac{\chi^2_J / J}{\chi^2_v / v}$$

Sean mis grados de libertad $=(n-k)$ entonces $t$ me queda:

$$t_{(n-k)} = \frac{Z}{\sqrt{\chi^2_{(n-k)} / (n-k)}}$$

Lo elevamos al cuadrado:

$$t^2_{(n-k)} = \left( \frac{Z}{\sqrt{\chi^2_{(n-k)} / (n-k)}} \right)^2$$

$$t^2_{(n-k)} = \frac{Z^2}{\left(\sqrt{\chi^2_{(n-k)} / (n-k)}\right)^2}$$

$$t^2_{(n-k)} = \frac{Z^2}{\chi^2_{(n-k)} / (n-k)}$$

Recprdemos que el cuadrado de una variable aleatoria normal estándar ($Z \sim N(0, 1)$) sigue una distribución Chi-cuadrado con 1 grado de libertad.Es decir: $Z^2 \sim \chi^2_1$

$$t^2_{(n-k)} = \frac{\chi^2_1}{\chi^2_{(n-k)} / (n-k)} → t^2_{(n-k)} = \frac{\chi^2_1/1}{\chi^2_{(n-k)} / (n-k)} =  \frac{\chi^2_J / J}{\chi^2_v / v} = F_{(J, v)}$$ **LQQD**
